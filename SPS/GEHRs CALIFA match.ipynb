```python
def finder(obj):
    path = os.getcwd()
    data = Table.read(path + '/CALIFA_serverlist.txt', format='ascii')

    Finder = 0
    
    for i in range(len(data)):
        if (obj in data['OBJ'][i]):
            data['Status'][i] = 1
            print(obj, '  Found')
            ascii.write(data, path + '/CALIFA_serverlist.txt', format='ecsv',overwrite = True)
            Finder = 1
            break
    if Finder == 1:
        print('\t')
        print('\t')
        print(data[data['Status'] == 1])
    else:
        print('Nothing found...')
```
