```python
import re
import os
import numpy as np
import pandas as pd
```

```python
tables_dir = os.path.join(os.path.expanduser("~"), 'HIIGalaxies/Distance_moduli/Compilation_tables')
```

```python
PLRC_tab_list = [f for f in os.listdir(tables_dir) if 'plrc' in f.lower()]
TRGB_tab_list = [f for f in os.listdir(tables_dir) if 'trgb' in f.lower()]
PLRC_tab_list.sort(key=lambda x: (int(x.split('.')[0]), x.split('.')[1]))
TRGB_tab_list.sort(key=lambda x: (int(x.split('.')[0]), x.split('.')[1]))
```

```python
PLRC_dfs_list = []

for i in range(len(PLRC_tab_list)):
    df = pd.read_csv(os.path.join(tables_dir,PLRC_tab_list[i]))
    df['galaxy'] = (PLRC_tab_list[i].split('.')[1]).replace('_PLRC','')

    columna_left = df.pop('galaxy')
    df.insert(0, 'galaxy', columna_left)

    df.rename(columns={
        'RefCode': 'bibcode',
        'Citation': 'citation',
        'Band': 'band',
        'mu_0': 'modulus',
        'e_R': 'random_error',
        'e_S': 'systematic_error',
        'e_T': 'total_error',
        'Category': 'category_error',
        'Ncef': 'cepheids_number',
        'Zcorr': 'metal_correction',
        'zeroP': 'zero_point',
        'Rank': 'probe_quality',
        'Year': 'publication_year',
        'Date': 'review_year',
        'Comments': 'comments',
        'Author': 'review_author',
        'ADS_date': 'ads_date',
        'JD': 'ads_jd',
    }, inplace=True)

    PLRC_dfs_list.append(df)


PLRC_final_DF = pd.concat(PLRC_dfs_list, ignore_index=True)
PLRC_final_DF.to_csv('PLRC_general_v1.1.csv',index=False)
PLRC_final_DF
```

```python
TRGB_dfs_list = []

for i in range(len(TRGB_tab_list)):
    df = pd.read_csv(os.path.join(tables_dir,TRGB_tab_list[i]))
    df['galaxy'] = (TRGB_tab_list[i].split('.')[1]).replace('_TRGB','')

    columna_left = df.pop('galaxy')
    df.insert(0, 'galaxy', columna_left)

    df.rename(columns={
        'RefCode': 'bibcode',
        'Citation': 'citation',
        'Band': 'band',
        'mu_0': 'modulus',
        'e_R': 'random_error',
        'Rank': 'probe_quality',
        'Year': 'publication_year',
        'Date': 'review_year',
        'Comments': 'comments',
        'Author': 'review_author',
        'ADS_date': 'ads_date',
        'JD': 'ads_jd',
    }, inplace=True)

    TRGB_dfs_list.append(df)


TRGB_final_DF = pd.concat(TRGB_dfs_list, ignore_index=True)
TRGB_final_DF.to_csv('TRGB_general_v1.1.csv',index=False)
TRGB_final_DF
```
